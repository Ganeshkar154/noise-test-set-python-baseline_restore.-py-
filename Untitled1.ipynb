```python
import argparse
import glob
import os
import cv2
import numpy as np
from skimage.restoration import denoise_tv_chambolle


def restore(arr: np.ndarray, scale: int = 4):
    """Denoise a noisy LR patch and upscale it.
    Returns (denoised_same_size, restored_upscaled).
    """
    x = np.clip(arr, 0, 1).astype(np.float32)
    # Total-variation denoising: edge-preserving, no extra deps (pywt-free).
    is_color = x.ndim == 3 and x.shape[-1] in (3, 4)
    den = denoise_tv_chambolle(x, weight=0.06, channel_axis=-1 if is_color else None)
    den = np.clip(den, 0, 1).astype(np.float32)

    # Cubic upscale + mild unsharp mask to recover perceived sharpness.
    up = cv2.resize(den, (den.shape[1] * scale, den.shape[0] * scale),
                     interpolation=cv2.INTER_CUBIC)
    blur = cv2.GaussianBlur(up, (0, 0), 1.0)
    sharp = np.clip(up + 0.6 * (up - blur), 0, 1)
    return den, sharp


def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--input", default="NoisyLR")
    ap.add_argument("--output", default="restored")
    ap.add_argument("--scale", type=int, default=4)

    # parse_known_args ignores extra/unrecognized args (e.g. Colab/Jupyter's
    # own "-f <kernel.json>") instead of raising SystemExit.
    args, _unknown = ap.parse_known_args()

    os.makedirs(args.output, exist_ok=True)
    files = sorted(glob.glob(os.path.join(args.input, "*.npy")))
    print(f"Found {len(files)} files")

    if not files:
        print(f"No .npy files found in '{args.input}/'. Check the --input path.")
        return

    for f in files:
        arr = np.load(f)
        _, restored = restore(arr, scale=args.scale)
        name = os.path.splitext(os.path.basename(f))[0]
        np.save(os.path.join(args.output, f"{name}_restored.npy"), restored)
        cv2.imwrite(os.path.join(args.output, f"{name}_restored.png"),
                     (restored * 255).astype(np.uint8))

    print(f"Done. Restored files written to {args.output}/")


if __name__ == "__main__":
    main()
```

```python
# ============================================
# STEP 4: Denoise / Restore the .npy Images
# ============================================

import numpy as np
import cv2
import glob
import os
import matplotlib.pyplot as plt

# Input and output folders
input_dir = "/content/Test_NoIsyLR/NoisyLR"
output_dir = "/content/Restored_Images"

# Get all .npy files
npy_files = sorted(glob.glob(os.path.join(input_dir, "*.npy")))

print("Total images:", len(npy_files))

# Process every image
for i, file_path in enumerate(npy_files):

    # Load .npy image
    img = np.load(file_path)

    # Convert to float32
    img = img.astype(np.float32)

    # Normalize to 0-255
    if img.max() > img.min():
        img = (img - img.min()) / (img.max() - img.min()) * 255.0
    else:
        img = np.zeros_like(img)

    img = img.astype(np.uint8)

    # Apply Gaussian denoising
    restored = cv2.GaussianBlur(img, (5, 5), 0)

    # Save as PNG
    filename = os.path.splitext(os.path.basename(file_path))[0]
    output_path = os.path.join(output_dir, filename + ".png")

    cv2.imwrite(output_path, restored)

    if (i + 1) % 100 == 0:
        print("Processed:", i + 1)

print("\nRestoration completed!")
print("Images saved in:", output_dir)
```

```python
# ============================================
# STEP 10: Calculate PSNR and SSIM for all images
# ============================================

import numpy as np
import cv2
import glob
import os
import pandas as pd
from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity

input_dir = "/content/Test_NoIsyLR/NoisyLR"
output_dir = "/content/Restored_Images"

npy_files = sorted(glob.glob(os.path.join(input_dir, "*.npy")))

results = []

for file_path in npy_files:

    # Load original
    original = np.load(file_path).astype(np.float32)

    # Normalize
    original = cv2.normalize(
        original, None, 0, 255, cv2.NORM_MINMAX
    ).astype(np.uint8)

    # Restored image
    name = os.path.splitext(os.path.basename(file_path))[0]
    restored_path = os.path.join(output_dir, name + ".png")

    restored = cv2.imread(
        restored_path,
        cv2.IMREAD_GRAYSCALE
    )

    if restored is None:
        continue

    # PSNR
    psnr = peak_signal_noise_ratio(
        original,
        restored,
        data_range=255
    )

    # SSIM
    ssim = structural_similarity(
        original,
        restored,
        data_range=255
    )

    results.append([name, psnr, ssim])

# Create table
df = pd.DataFrame(
    results,
    columns=["Image", "PSNR (dB)", "SSIM"]
)

print(df)

print("\nAverage PSNR:",
      df["PSNR (dB)"].mean(), "dB")

print("Average SSIM:",
      df["SSIM"].mean())
```

```python
# ============================================
# STEP 12: CNN Image Denoising Model
# ============================================

import os
import glob
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# -----------------------------
# Settings
# -----------------------------
input_dir = "/content/Test_NoIsyLR/NoisyLR"
model_path = "/content/cnn_denoising_model.keras"

# Load .npy files
files = sorted(glob.glob(os.path.join(input_dir, "*.npy")))

print("Total images:", len(files))

# -----------------------------
# Load images
# -----------------------------
images = []

for file in files:
    img = np.load(file).astype(np.float32)

    # Normalize to 0-1
    img_min = img.min()
    img_max = img.max()

    if img_max > img_min:
        img = (img - img_min) / (img_max - img_min)
    else:
        img = np.zeros_like(img)

    # Add channel dimension
    img = img[..., np.newaxis]

    images.append(img)

images = np.array(images, dtype=np.float32)

print("Dataset shape:", images.shape)

# -----------------------------
# Train / validation split
# -----------------------------
np.random.seed(42)

indices = np.random.permutation(len(images))

split = int(0.8 * len(images))

train_images = images[indices[:split]]
val_images = images[indices[split:]]

print("Training images:", len(train_images))
print("Validation images:", len(val_images))

# -----------------------------
# Create synthetic noisy input
# -----------------------------
def add_noise(x):
    noise = np.random.normal(
        loc=0.0,
        scale=0.08,
        size=x.shape
    ).astype(np.float32)

    noisy = x + noise

    return np.clip(noisy, 0.0, 1.0)

# Create training inputs
train_noisy = add_noise(train_images)

val_noisy = add_noise(val_images)

# -----------------------------
# Build CNN
# -----------------------------
model = models.Sequential([

    layers.Input(shape=(128, 128, 1)),

    layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        64, (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        64, (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        1, (3, 3),
        padding="same",
        activation="sigmoid"
    )
])

# -----------------------------
# Compile
# -----------------------------
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# -----------------------------
# Train
# -----------------------------
history = model.fit(
    train_noisy,
    train_images,
    validation_data=(val_noisy, val_images),
    epochs=10,
    batch_size=16,
    shuffle=True
)

# -----------------------------
# Save model
# -----------------------------
model.save(model_path)

print("\nCNN training completed!")
print("Model saved to:", model_path)
```

```python
# ============================================
# STEP 13: Test CNN Restoration
# ============================================

# Select one validation image
test_input = val_noisy[0:1]
test_target = val_images[0]

# Predict restored image
prediction = model.predict(test_input, verbose=0)[0]

# Remove channel dimension
prediction = prediction.squeeze()
test_input_display = test_input[0].squeeze()

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(test_input_display, cmap="gray")
plt.title("Noisy Input")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(prediction, cmap="gray")
plt.title("CNN Restored")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(test_target.squeeze(), cmap="gray")
plt.title("Target")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# STEP 14: CNN PSNR and SSIM
# ============================================

from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity

target = test_target.squeeze()

psnr_cnn = peak_signal_noise_ratio(
    target,
    prediction,
    data_range=1.0
)

ssim_cnn = structural_similarity(
    target,
    prediction,
    data_range=1.0
)

print("CNN PSNR :", psnr_cnn, "dB")
print("CNN SSIM :", ssim_cnn)
```

```python
# ============================================
# STEP 16: Restore Images Using Trained CNN
# ============================================

import os
import glob
import numpy as np
import cv2
import tensorflow as tf
from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity

# Load trained model
model_path = "/content/cnn_denoising_model.keras"
model = tf.keras.models.load_model(model_path)

# Input and output folders
input_dir = "/content/Test_NoIsyLR/NoisyLR"
cnn_output_dir = "/content/CNN_Restored_Images"

os.makedirs(cnn_output_dir, exist_ok=True)

# Find all NPY files
npy_files = sorted(
    glob.glob(os.path.join(input_dir, "*.npy"))
)

print("Total images:", len(npy_files))

psnr_values = []
ssim_values = []

# Process images
for i, file_path in enumerate(npy_files):

    # Load image
    img = np.load(file_path).astype(np.float32)

    # Normalize to 0-1
    min_val = img.min()
    max_val = img.max()

    if max_val > min_val:
        img_norm = (img - min_val) / (max_val - min_val)
    else:
        img_norm = np.zeros_like(img)

    # Create synthetic noisy input
    noise = np.random.normal(
        0, 0.08, img_norm.shape
    ).astype(np.float32)

    noisy = np.clip(
        img_norm + noise,
        0,
        1
    )

    # Add batch and channel dimensions
    model_input = noisy[np.newaxis, ..., np.newaxis]

    # CNN prediction
    restored = model.predict(
        model_input,
        verbose=0
    )[0, ..., 0]

    # Calculate PSNR
    psnr = peak_signal_noise_ratio(
        img_norm,
        restored,
        data_range=1.0
    )

    # Calculate SSIM
    ssim = structural_similarity(
        img_norm,
        restored,
        data_range=1.0
    )

    psnr_values.append(psnr)
    ssim_values.append(ssim)

    # Convert restored image to 0-255
    restored_uint8 = (
        np.clip(restored, 0, 1) * 255
    ).astype(np.uint8)

    # Save image
    name = os.path.splitext(
        os.path.basename(file_path)
    )[0]

    save_path = os.path.join(
        cnn_output_dir,
        name + ".png"
    )

    cv2.imwrite(
        save_path,
        restored_uint8
    )

    if (i + 1) % 50 == 0:
        print("Processed:", i + 1)

# ============================================
# Final Results
# ============================================

print("\n================================")
print("CNN RESTORATION COMPLETED")
print("================================")

print("Average PSNR:",
      np.mean(psnr_values), "dB")

print("Average SSIM:",
      np.mean(ssim_values))

print("Output folder:")
print(cnn_output_dir)
```
